```python
# 594 -Longest Harmonious Subsequence 

class Solution(object):
    def findLHS(self, nums):
        """
        :type nums: List[int]
        :rtype: int
        """
        count={}
        # count of each elements
        for num in nums:
            count[num]=count.get(num,0) + 1
        longest=0
        for num in nums:
            if num+1 in count:
                length=count[num] +  count[num+1]
                longest= max(length,longest)
        return longest
```
